# Informasi Dataset

Jumlah kelas: 4

Kategori:

1. Kaca
2. Kertas
3. Logam
4. Plastik

Preprocessing:

- Auto-Orient
- Resize 640x640

Augmentasi:

- Horizontal Flip
- Vertical Flip
- Rotate 90°
- Upside Down
- Crop maksimum 20%

```python
# Load Model Dataset YOLOv8 (kalau mau latih yolov8s tinggal ganti modelnya)

model = YOLO("yolov8n.pt")
```

# Parameter Training

Model: YOLOv8n

Epoch: 60

Image Size: 640

Batch Size: 12

Workers: 4

Device: GPU (CUDA)

Cache: False

```python
# Training Model

results = model.train(
    data="data.yaml",
    epochs=60,
    imgsz=640,
    batch=12,
    workers=4,
    cache=False,
    device=0
)

# Catatan
# Training dapat memakan waktu 2–4 jam pada RTX 4050 Laptop GPU.
```

```python
# Evaluasi Model

metrics = model.val(
    plots=True
)

print(metrics)
```

```python
# Menampilkan Hasil Evaluasi

print("Precision:", metrics.box.mp)
print("Recall:", metrics.box.mr)
print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
```

```python
# Prediksi Menggunakan Model

model = YOLO("runs/detect/train/weights/best.pt")

results = model.predict(
    source="dataset/test/images",
    save=True
)
```
