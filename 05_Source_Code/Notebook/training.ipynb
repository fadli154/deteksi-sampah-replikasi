## 🔍 2. Verifikasi GPU (CUDA)
Sangat disarankan menggunakan GPU NVIDIA (atau GPU T4 di Google Colab) agar proses training berjalan cepat.

```python
import torch

print("PyTorch Version:", torch.__version__)
print("CUDA Available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU Device Name:", torch.cuda.get_device_name(0))
else:
    print("⚠️ Peringatan: CUDA tidak aktif. Training dengan CPU akan memakan waktu > 12 jam!")
```

## 📋 3. Konfigurasi Parameter Pelatihan
Tentukan parameter training Anda di bawah ini. Anda dapat memilih tipe model `yolov8n` (Nano - ringan) atau `yolov8s` (Small - lebih akurat).

```python
# Parameter Konfigurasi Pelatihan
model_variant = "yolov8n"  # Pilihan: 'yolov8n' atau 'yolov8s'
epochs = 60
batch_size = 12
imgsz = 640
device = 0 if torch.cuda.is_available() else 'cpu'

print(f"Menyiapkan training untuk {model_variant}...")
print(f"Epochs: {epochs}, Batch Size: {batch_size}, Image Size: {imgsz}, Device: {device}")
```

## 🏋️‍♂️ 4. Jalankan Proses Training
Cell ini akan mengunduh model dasar terlebih dahulu, kemudian memulai training selama 60 epoch. Jika Anda kehabisan memori GPU (Out of Memory/OOM), silakan kurangi `batch_size` ke 8 atau 4.

```python
from ultralytics import YOLO

# Muat model pre-trained
model = YOLO(f"{model_variant}.pt")

# Jalankan training
results = model.train(
    data="data.yaml",
    epochs=epochs,
    imgsz=imgsz,
    batch=batch_size,
    workers=4,
    cache=False,
    device=device,
    name=f"train-{model_variant}"
)

print("\n🎉 Proses training berhasil diselesaikan!")
print("Hasil tersimpan di:", results.save_dir)
```
