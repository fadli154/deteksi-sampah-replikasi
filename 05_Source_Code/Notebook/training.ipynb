# 🚀 Tahap 2: Pelatihan Model (Training) YOLOv8

**Proyek Replikasi Penelitian YOLOv8**  
Dataset: Recyclable Waste (Roboflow)

### Deskripsi Tahap Ini:
Pada notebook ini, kita akan melakukan proses pelatihan model (training) menggunakan **YOLOv8**:
1. Memeriksa ketersediaan GPU (NVIDIA CUDA) untuk mempercepat proses pelatihan.
2. Mengunduh bobot awal pre-trained model YOLOv8 (yolov8n.pt / yolov8s.pt).
3. Menjalankan proses training sebanyak **60 epoch** sesuai dengan parameter replikasi penelitian.
4. Menyimpan hasil pelatihan secara otomatis ke folder `runs/detect/`.

### 🛠️ 1. Setup Environment dan Verifikasi GPU
Jalankan cell di bawah ini untuk menginstal framework `ultralytics` (YOLO) dan memastikan GPU Nvidia Anda terdeteksi dengan CUDA aktif.

```python
# !pip install ultralytics torch torchvision

import torch
from ultralytics import YOLO
import os

print("PyTorch Version:", torch.__version__)
print("CUDA Available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU Device Name:", torch.cuda.get_device_name(0))
else:
    print("⚠️ Peringatan: CUDA tidak aktif. Training dengan CPU akan memakan waktu > 12 jam!")
```

### 📋 2. Konfigurasi Parameter Pelatihan
Ubah model ke `yolov8n.pt` (untuk Nano model yang ringan) atau `yolov8s.pt` (untuk Small model yang lebih akurat namun sedikit lebih berat).

```python
# Parameter Konfigurasi Pelatihan
model_variant = "yolov8n"  # Pilihan: 'yolov8n' atau 'yolov8s'
epochs = 60
batch_size = 12
imgsz = 640
device = 0 if torch.cuda.is_available() else 'cpu'

notebook_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(notebook_dir, "..", ".."))
yaml_path = os.path.join(project_root, "data.yaml")

print(f"Menyiapkan training {model_variant}...")
print(f"Menggunakan dataset: {yaml_path}")
```

### 🏋️‍♂️ 3. Jalankan Proses Training
Proses ini akan mengunduh model dasar terlebih dahulu, kemudian memulai training selama 60 epoch. Jika Anda menggunakan laptop GPU RTX 4050, proses ini memakan waktu kurang lebih **2-4 jam**.

```python
# Muat model pre-trained
model = YOLO(f"{model_variant}.pt")

# Eksekusi pelatihan
try:
    results = model.train(
        data=yaml_path,
        epochs=epochs,
        imgsz=imgsz,
        batch=batch_size,
        workers=4,
        cache=False,
        device=device,
        name=f"train-{model_variant}"
    )
    print("\n🎉 Proses training berhasil diselesaikan!")
    print("Hasil tersimpan di:", results.save_dir)
except Exception as e:
    print("❌ Error dalam training:", e)
    print("Jika CUDA kehabisan memori, coba kecilkan batch_size ke 8 atau 4 pada cell sebelumnya.")
```
